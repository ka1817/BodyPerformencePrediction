```python
import pandas as pd
import scipy.stats as stats

# Group ages by class
grouped_ages = [df[df['class'] == c]['age'] for c in df['class'].unique()]

# Perform ANOVA test
f_stat, p_value = stats.f_oneway(*grouped_ages)

# Print Results
print(f"F-statistic: {f_stat:.3f}, P-value: {p_value:.3f}")

# Interpretation
if p_value < 0.05:
    print("There is a significant difference in age across classes.")
else:
    print("There is no significant difference in age across classes.")
```

Since the P-value is 0.000 (which is less than 0.05), we reject the null hypothesis. This means that there is a significant difference in age across the class groups.

```python
from scipy.stats import chi2_contingency

contingency_table = pd.crosstab(df['gender'], df['class'])
chi2, p, dof, expected = chi2_contingency(contingency_table)

print(f"Chi-square Statistic: {chi2:.3f}")
print(f"P-value: {p:.5f}")

if p < 0.05:
    print("There is a significant relationship between Gender and Class.")
else:
    print("No significant relationship between Gender and Class.")
```

```python
import pandas as pd
import scipy.stats as stats

# Select numerical features (excluding 'class')
numerical_features = df.select_dtypes(include=['number']).columns

# Dictionary to store ANOVA results
anova_results = {}

# Perform ANOVA for each numerical feature
for feature in numerical_features:
    groups = [df[df['class'] == cls][feature] for cls in df['class'].unique()]
    f_stat, p_value = stats.f_oneway(*groups)
    anova_results[feature] = {'F-statistic': f_stat, 'P-value': p_value}

# Convert results to DataFrame for better visualization
anova_df = pd.DataFrame(anova_results).T

# Display results
print(anova_df)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline 
df['gender']=df['gender'].map({'M':1,'F':0})
```

```python
X=df.drop('class',axis=1)
y=df['class']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
X_train
```

```python
#scaling
from sklearn.preprocessing import StandardScaler
trf1=ColumnTransformer([
    ('scale',StandardScaler(),slice(0,7))
])
```

```python
from sklearn.linear_model import LogisticRegression
trf2=LogisticRegression()
```

```python
pipe=Pipeline([
    ('trf1',trf1),
    ('trf2',trf2),
])
```

```python
pipe.fit(X_train,y_train)
```

```python
y_pred=pipe.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```
